```python
import numpy as np
import matplotlib.pyplot as plt
```

### Cell values

```python
params_cell = np.loadtxt('params_eta_gamma_cell_v=3_L=10_r=1.txt')
reward_cell = np.loadtxt('mean_reward_cell_v=3_L=10_r=1.txt')
```

```python
eta_c = params_cell[0]
gamma_c = params_cell[1]
```

```python
plt.matshow(reward_cell)

plt.xlabel('eta')
plt.ylabel('gamma')
plt.xticks(ticks=np.arange(len(eta_c)), labels=[f"{e:.2g}" for e in eta_c], rotation=45)
plt.yticks(ticks=np.arange(len(gamma_c)), labels=[f"{g:.1g}" for g in gamma_c])
plt.colorbar(label ='Average reward over last 500 episodes')
```
